# Slutsats

## Vad kom vi fram till?

Både tid och väder hjälper oss att förutsäga hur många cyklar som hyrs ut när systemet är öppet. Vi såg tydliga skillnader över dygnet, men också mellan varma och kalla timmar och mellan timmar med och utan regn.

**Random Forest gav lägst MAE när modellerna fick samma väder- och tidsinformation och testades på samma slumpmässigt utvalda timmar.** Sedan undersökte vi modellens inställningar och vilken information den skulle använda. Då delade vi upp datan i tidsordning och använde en senare period för validering.

---

## Hur bra blev resultatet?

| Mått | Slutlig Random Forest – kronologiskt test |
|---|---:|
| MAE | **261,02 uthyrningar per timme** |
| RMSE | **352,94 uthyrningar per timme** |
| R² | **0,592** |

Den **kronologiska baslinjen** missade i genomsnitt med ungefär **444 uthyrningar per timme**. Slutmodellen missade med ungefär **261**. Det är cirka **41 % lägre MAE** på samma testperiod.

Baslinjen på **509,7** i den tidigare modelljämförelsen använde en annan, slumpmässig testuppdelning. Den ska därför inte jämföras direkt med slutmodellens resultat.

Modellen följde dygnsmönstret, men gissade ofta för lågt när uthyrningen var som högst. RMSE ger dessa stora missar större tyngd. R² visar hur väl modellen fångar variationen mellan timmarna.

---

## Vad tar vi med oss?

**Hur vi delar upp datan spelar roll.** Vi valde att träna på tidigare data och använda senare data för validering och test, med fördelningen 70/15/15. Då tränas modellen inte på timmar som inträffar efter testperioden.

MAE var ungefär 181 på valideringen och 261 på sluttestet. Ett bra resultat under en period betyder alltså inte att modellen fungerar lika bra under nästa.

Vi har bara ett års data från Seoul och använder uppmätt väder. Vi hade också redan undersökt delar av den senare testperioden under arbetet. Därför behövs fler tester på ny data innan vi vet hur bra modellen fungerar i praktiken.

Vi kan ännu inte säga säkert om tid eller väder ensamt hjälper mest. Ett nästa steg är att jämföra tid, väder och kombinationen med samma modell och samma uppdelning.

**Det viktigaste vi tar med oss är att inte bara titta på hur lågt felet är, utan också på hur modellen har testats.**
